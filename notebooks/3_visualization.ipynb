```bash
python scripts/run_all.py --config configs/eurosat_benchmark_all.yaml --skip_existing
```

The cells below focus only on the model run produced by `2_model.ipynb`. By default, that run is read from `outputs/eurosat/latest`, and its visualizations are written to `paper_assets/analytics/latest/`.


## Select the Run From `2_model.ipynb`


````python
from pathlib import Path
import subprocess
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

PYTHON = sys.executable
OUTPUTS_ROOT = PROJECT_ROOT / "outputs" / "eurosat"
PAPER_ASSETS = PROJECT_ROOT / "paper_assets"

# `2_model.ipynb` refreshes this folder when it saves the selected run.
MODEL_RUN = OUTPUTS_ROOT / "latest"
ANALYTICS_ROOT = PAPER_ASSETS / "analytics"
MODEL_RUN
````

## Generate Visualizations for the Selected Run


````python
if not (MODEL_RUN / "metrics.json").exists():
    raise FileNotFoundError(
        f"No completed run found at {MODEL_RUN}. Run 2_model.ipynb first, or set MODEL_RUN to a timestamped folder under outputs/eurosat."
    )

subprocess.run(
    [PYTHON, "scripts/run_analytics.py", str(MODEL_RUN), "--output_dir", str(ANALYTICS_ROOT)],
    cwd=PROJECT_ROOT,
    check=True,
)

analytics_dir = ANALYTICS_ROOT / MODEL_RUN.name
analytics_dir
````

## Generated Figures


````python
from IPython.display import Markdown, display

pdfs = sorted(analytics_dir.glob("*.pdf"))
if not pdfs:
    display(Markdown("No PDF figures were generated for this run."))
else:
    display(Markdown(f"Generated `{len(pdfs)}` figures in `{analytics_dir.relative_to(PROJECT_ROOT)}`:"))
    for path in pdfs:
        display(Markdown(f"- `{path.relative_to(PROJECT_ROOT)}`"))
````

## Summary


````python
summary_path = analytics_dir / "analytics_summary.md"
display(Markdown(summary_path.read_text(encoding="utf-8") if summary_path.exists() else "No analytics summary found."))
````
